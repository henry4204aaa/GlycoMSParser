```python
#modified from programmiz
def binarycompositionsearch(array, x, low, high):
    # Repeat until the pointers low and high meet each other
    while low <= high:
        mid = low + (high - low)//2
        print(f"mid is {mid} now")
        print(f"value:{array[mid]-x}")
        if abs(array[mid])-x)<5:
            return mid
        elif array[mid] < x:
            print("mid<x")
            low = mid + 1
            print(f"now lower bound is {low} and it's {array[low]}")
        else:
            print("mid>x")
            high = mid - 1
            print(f"now higher bound is {high} and it's {array[high]}")
    return -1

array = sortcomp[1]
array1 = [3, 4, 5, 6, 7, 8, 9]
x = 1714

result = binarycompositionsearch(array, x, 0, len(array)-1)

if result != -1:
    print("Element is present at index " + str(result))
else:
    print("Not found")
```
