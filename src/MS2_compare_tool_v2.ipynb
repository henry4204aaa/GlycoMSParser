```python
#基本輸入輸出詳情範例參考MS2_compare_tool_v0
import pandas as pd
import re
file=pd.read_csv('zf_sPerMeNG_brain MS2 summary from 44218 at 20230307-153550.csv',sep='\t')
print(file["peaklist"])
print(type(file["peaklist"][1]))

MSinput=[]
for i in range(len(file["peaklist"])): 
    match = re.search(r'MS2peaklist\(dMass=\((.*?)\)',file["peaklist"][i] )
    if match:
        # 取出所有浮點數
        ms2peaklist_str = match.group(1)
        #ms2peaklist = [round(float(x),2) for x in ms2peaklist_str.split(',')]##
        ms2peaklist = [round(float(x),4) for x in ms2peaklist_str.split(',')]
        #print(ms2peaklist)
        MSinput.append(ms2peaklist)
    else:
        print('not match')
        break
print(len(MSinput))


def creat_Gdic(Gfrag,Gname):
    Gdic={}
    for i in range(len(Gname)):
        Gdic[Gname[i]]=Gfrag[i]

    return Gdic
def creat_Mdic(Gfrag,Gname):
    Mdic={}
    
    for n in range(len(Gfrag)):
        for t in Gfrag[n]:
            li=[]
            for i in range(len(Gfrag)):
                if t in Gfrag[i]:
                    #print(Gfrag[i])
                    li.append(Gname[i])
                    #print(li)
            Mdic[t]=li
    return Mdic

def creat_GFCount(gname,Gfrag):
    GFCount={}
    for i in range(len(gname)):
        GFCount[gname[i]]=len(Gfrag[i])
    return GFCount



```

```python
##原版
def creat_Td(M_list,Mdic):
    Td={}
    for m in M_list:
        if m in Mdic:
            for T in Mdic[m]:
                if T in Td:
                    Td[T].append(m)
                else:
                    Td[T]=[m]
        else: continue
    return Td
creat_Td([0,1,2],{2: ['G1', 'G2', 'G3'], 3: ['G2'], 4: ['G3']})    
```

```python
def glycancomposition(a, b, c, d, e):
    M =  a * 174.08921 + b * 204.09977 + c * 245.12632 + d * 361.17367 + e * 391.18423 + 46.04186 + 1.0073
    # M = a x Fucose (triangle) + b x Hexose (circle) + c x HexNAc (Square) + d * Neu5Ac (purple diamond) + e * Neu5Gc (aqua diamond) + reducing end + [H+]
    # a, b, c, d, e >=0, is integer
    return(M)
glycancomposition(1, 5, 4, 2, 0)
```
